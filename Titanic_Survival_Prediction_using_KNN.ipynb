```python
import pandas as pd
from urllib.request import urlretrieve
from pathlib import Path
```

```python
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

train_url = "https://raw.githubusercontent.com/agconti/kaggle-titanic/master/data/train.csv"
test_url = "https://raw.githubusercontent.com/agconti/kaggle-titanic/master/data/test.csv"

urlretrieve(train_url, data_dir / "train.csv")
urlretrieve(test_url, data_dir / "test.csv")

print("Datasets downloaded successfully!")
```

```python
train_df = pd.read_csv("data/train.csv")
test_df = pd.read_csv("data/test.csv")

print("Train data:")
display(train_df.head())

print("Test data:")
display(test_df.head())
```

```python
train_df = pd.read_csv("data/train.csv")
test_df = pd.read_csv("data/test.csv")

print("Train data:")
display(train_df.head())

print("Test data:")
display(test_df.head())
```

```python
print("Train missing values:")
print(train_df.isna().sum())

print("\nTest missing values:")
print(test_df.isna().sum())
```

```python
passenger_id_df = test_df['PassengerId']

train_df = train_df.drop(
    columns=['Cabin', 'PassengerId', 'Name', 'Ticket']
)

test_df = test_df.drop(
    columns=['Cabin', 'PassengerId', 'Name', 'Ticket']
)

print("Columns after dropping:")
print(train_df.columns)
```

```python
age_mean = train_df['Age'].mean()

print("Mean Age:", age_mean)

train_df['Age'] = train_df['Age'].fillna(age_mean)
test_df['Age'] = test_df['Age'].fillna(age_mean)

print("Missing Age in train:", train_df['Age'].isna().sum())
print("Missing Age in test:", test_df['Age'].isna().sum())
```

```python
fare_mean = train_df['Fare'].mean()

print("Mean Fare:", fare_mean)

test_df['Fare'] = test_df['Fare'].fillna(fare_mean)

print("Missing Fare in test:", test_df['Fare'].isna().sum())
```

```python
embarked_mode = train_df['Embarked'].mode()[0]

print("Embarked mode:", embarked_mode)

train_df['Embarked'] = train_df['Embarked'].fillna(embarked_mode)

print("Missing Embarked:", train_df['Embarked'].isna().sum())
```

```python
print("Train missing values:")
print(train_df.isna().sum())

print("\nTest missing values:")
print(test_df.isna().sum())
```

```python
train_df = pd.get_dummies(train_df)
test_df = pd.get_dummies(test_df)

print("Train columns:")
print(train_df.columns)

print("\nTest columns:")
print(test_df.columns)
```

```python
target = train_df.pop('Survived')

print("Target:")
display(target.head())

print("\nTraining features:")
display(train_df.head())
```

```python
test_df = test_df.reindex(
    columns=train_df.columns,
    fill_value=0
)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

train_df = scaler.fit_transform(train_df)
test_df = scaler.transform(test_df)

print("Feature scaling completed!")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
```

```python
knn = KNeighborsClassifier()

print(knn)
```

```python
param_grid = {
    'n_neighbors': range(1, 21),
    'p': [1, 2]
}

param_grid
```

```python
grid_search = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)

grid_search.fit(train_df, target)

print("Grid Search completed!")
```

```python
print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation accuracy:")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

print(best_model)
```

```python
best_model.fit(train_df, target)

print("Best KNN model trained successfully!")
```

```python
predicted_test_Y = best_model.predict(test_df)

print(predicted_test_Y)
```

```python
predicted_test_Y_df = pd.DataFrame()

predicted_test_Y_df['PassengerId'] = passenger_id_df
predicted_test_Y_df['Survived'] = predicted_test_Y

display(predicted_test_Y_df.head(10))
```

```python
predicted_test_Y_df.to_csv(
    "titanic_knn_submission.csv",
    index=False,
    header=True
)

print("Submission file created successfully!")
```
